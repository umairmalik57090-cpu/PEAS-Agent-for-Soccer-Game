```python
# Soccer Game Agent using PEAS Model

class SoccerAgent:
    def __init__(self):
        self.ball_near = False
        self.opponent_near = False

    def sense_environment(self, ball_near, opponent_near):
        self.ball_near = ball_near
        self.opponent_near = opponent_near
        print("Sensing environment...")
        print(f"Ball Near: {self.ball_near}")
        print(f"Opponent Near: {self.opponent_near}")

    def decide_action(self):
        if self.ball_near:
            if self.opponent_near:
                return "Pass the ball"
            else:
                return "Shoot towards goal"
        else:
            return "Move towards the ball"

    def act(self, action):
        print(f"Action taken: {action}")


# Main Program
agent = SoccerAgent()

# Simulated environment input
agent.sense_environment(ball_near=True, opponent_near=False)
action = agent.decide_action()
agent.act(action)
```
